```python

# Student Performance Analysis System
# Using Lists, Dictionaries, Map and Sets

students = [
    {
        "name": "John",
        "age": 20,
        "address": "New York",
        "marks": [85, 90, 78, 92, 88]
    },
    {
        "name": "jack",
        "age": 21,
        "address": "New Jersey",
        "marks": [75, 80, 85, 70, 90]
    },
    {
        "name": "David",
        "age": 19,
        "address": "California",
        "marks": [90, 95, 88, 92, 96]
    },
    {
        "name": "Sandy",
        "age": 22,
        "address": "Texas",
        "marks": [65, 72, 68, 75, 70]
    }
]

print("STUDENT PERFORMANCE ANALYSIS SYSTEM")


# LIST: Store all student records
# DICTIONARY: Store individual student information
# MAP: Calculate total marks for every student

total_marks = list(
    map(lambda student: sum(student["marks"]), students)
)

# MAP: Calculate average marks for every student

average_marks = list(
    map(
        lambda student: sum(student["marks"]) / len(student["marks"]),
        students
    )
)

# SET: Identify unique student addresses

unique_addresses = set(
    map(lambda student: student["address"], students)
)

# SET: Identify unique student ages

unique_ages = set(
    map(lambda student: student["age"], students)
)

# Display student information

for i, student in enumerate(students):

    print("\nStudent Name:", student["name"])
    print("Age:", student["age"])
    print("Address:", student["address"])
    print("Marks:", student["marks"])
    print("Total Marks:", total_marks[i])
    print("Average Marks:", round(average_marks[i], 2))



# Overall class analysis

class_total = sum(total_marks)
class_average = sum(average_marks) / len(students)

# Find topper using dictionary data

topper_index = average_marks.index(max(average_marks))
topper = students[topper_index]

print("\nCLASS PERFORMANCE SUMMARY")


print("Number of Students:", len(students))
print("Total Class Marks:", class_total)
print("Class Average:", round(class_average, 2))

print("Topper:", topper["name"])
print("Highest Average:", round(max(average_marks), 2))

print("Unique Addresses:", unique_addresses)
print("Unique Ages:", unique_ages)

```
